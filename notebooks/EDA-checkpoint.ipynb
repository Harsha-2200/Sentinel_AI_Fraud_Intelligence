```python
import pandas as pd

df = pd.read_parquet('sentinel_features_cleaned.parquet')


#Derieving hour and day from TransactionDT
df['hour'] = (df['TransactionDT']//3600) % 24
print(df[['hour']])
print()
df['day'] = (df['TransactionDT']//86400)
print(df[['day']])
# Both hour and day are created from TransactionDT
```

```python
#Cheking overall fraud percent
fraud_pct = (df['isFraud'].mean()) * 100
print(fraud_pct)                 # overall fraud in dataset is 3.5%
```

```python
#Cheking TransactionAmt distribution
print("TransactionAmt description:")
print(df['TransactionAmt'].describe())
print()
print("Cheking Skewness:")
print(df['TransactionAmt'].skew())
# The skewness is 14.37 for the entire dataset. The 25th percentile is 43.32 and the 75th percentile is 125.00.
# while most values sit in this relatively small range, the maximum transaction amount reaches 31,937.39. 
# The minimum amount is just 0.25 while the highest is 31,937.39, showing the values are highly skewed and create real outliers at the upper end.
```

```python
#Cheking IQR and outliers
Q1 = df['TransactionAmt'].quantile(0.25)
Q3 = df['TransactionAmt'].quantile(0.75)
IQR = Q3 - Q1
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

outliers = df[(df['TransactionAmt'] < lower_fence) | (df['TransactionAmt'] > upper_fence)]
print(f"Q1: {Q1}\nQ3: {Q3}\nIQR: {IQR}")
print(f"Lower fence: {lower_fence}\nUpper fence: {upper_fence}")
print(f"Number of outliers: {len(outliers)}")
print(f"Percentage of dataset: {len(outliers)/len(df)*100:.2f}%")

# Fraud rate specifically within the outlier group vs overall
print(f"\nFraud rate among outliers: {outliers['isFraud'].mean()*100:.2f}%")
print(f"Overall fraud rate: {df['isFraud'].mean()*100:.2f}%")

# Using the IQR method, any transaction below -79.20 or above 247.52 is considered a statistical outlier. 
# Since amounts can't be negative, the real cutoff that matters is 247.52.
# 66,482 transactions (about 11.3% of the full dataset) fall above this upper limit, confirming the heavy skew already found in the distribution check.
# These are not necessarily fraud — they represent unusually large purchases, which can be genuine 
# (like real customers making bigger purchases) or suspicious, depending on other factors.
# Comparing fraud rate inside this outlier group versus the overall baseline shows whether large transactions 
# are more or less risky than typical ones, which is useful information for setting risk thresholds later.
```

```python
#Graphical representation of TransactionAmt
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data=df, x='TransactionAmt', kde=True, ax=axes[0])
axes[0].set_title("TransactionAmt Distribution")
axes[0].set_xlabel("Amount")

sns.boxplot(data=df, y='TransactionAmt', ax=axes[1])
axes[1].set_title("TransactionAmt Outliers")

plt.tight_layout()
plt.savefig('sentinel_transactionamt_dist.png', dpi=150)
plt.show()

# The histogram confirms the extreme right-skew already found numerically (skewness 14.37) — 
# nearly all transactions cluster near zero, with a long, barely visible tail stretching out to over 30,000.
# The boxplot shows the same pattern visually — most values are compressed into a thin band near the bottom, 
# with a dense cluster of outlier points above the upper fence, including one extreme transaction near 32,000.
# This confirms transaction amount cannot be used as a simple threshold-based fraud indicator on its own — 
# the distribution is too skewed, and both genuine and fraudulent transactions likely span a wide range of amounts.
```

```python
#Cheking Card1 transaction distribution
card_counts = df['card1'].value_counts()
print(type(card_counts))
print(card_counts.head(10))
print(f"\nNumber of distinct card1 values: {len(card_counts)}")
print(card_counts.describe())

#There are 13535 distinct cards in card1 column. We can see in median(50%) there are 4 transactions per card 
# and mean is 43.57 which is extream compared to mean.
```

```python
#Divinding card1 into buckets
bins = [0,1,10,50,500,20000]
labels=['1_transaction', '2_to_10', '11_to_50', '51_to_500', '500_plus']
card_counts_binned = pd.cut(card_counts,bins=bins,labels=labels)
print(card_counts_binned.value_counts().sort_index())

# Checked how many transactions each unique card1 value has across the full dataset, rather than assuming card1 always represents one individual card.
# Most cards behave normally, 45% have between 2 and 10 transactions, and 25% appear only once in the entire dataset.
# A small group of cards (about 1.5%, or 199 out of 13,553) show extremely high transaction counts, some in the thousands
# far beyond what one individual customer could realistically generate.
# These high-volume cards likely represent shared, default, or category-level codes rather than genuine individual cards, 
# and should be handled separately in feature engineering (such as velocity detection) rather than treated the same as normal,individual-card behavior.
```

```python
# Insight — Card1 Behavior Investigation

# Checked whether card1 always represents one individual card before using it to build features like velocity detection. 
# Looking at the most frequent card1 values, several (like 7919, 9500, 15885) showed thousands of transactions each 
# far more than one real person could realistically make.

# To see how common this was, the full spread of transaction counts per card was checked. Most cards look normal
# 70% have 10 or fewer transactions, matching typical customer behavior. But a small group, about 1.5% of all cards (199 out of 13,553), 
# show extremely high transaction counts, some reaching into the thousands.

# What this means: card1 doesn't always represent a single real card. It seems to be a mix, mostly real individual cards, 
# but with a small number of shared or default codes mixed in that behave very differently. Any feature built using card1, 
# especially something like velocity (how many transactions happen close together), needs to treat this small group separately, 
# or it will add noise instead of real fraud signal.
```

```python
#Correlation Matrix between isFraud and TransactionAmt,TransactionDT,hour,day,card1.
print("Correlation Matrix between isFraud and TransactionAmt:")
print(df['isFraud'].corr(df['TransactionAmt']))

print("\nCorrelation Matrix between isFraud and TransactionDT:")
print(df['isFraud'].corr(df['TransactionDT']))

print("\nCorrelation Matrix between isFraud and card1:")
print(df['isFraud'].corr(df['card1']))

print("\nCorrelation Matrix between isFraud and hour:")
print(df['isFraud'].corr(df['hour']))

print("\nCorrelation Matrix between isFraud and day:")
print(df['isFraud'].corr(df['day']))

#We can see correlation is very weak between isFraud and numeric columns(TransactionAmt,TransactionDT,hour,day,card1).this is because 
#the dataset is driven by non-linear, categorical patterns (like specific product types or specific hour windows) rather than simple linear trends
```

```python
#Checking hour of day fraud rate.
print(df.groupby('hour')['isFraud'].mean()*100)

#We can see that fraud rate as peaked at hoour 7 and the fraud rate drops after it. The peak hours for fraud is between 5-9.
#Most of the hours the fraud rate is in between 2%-4%.
```

```python
# EDA Summary — Key Findings

# Fraud makes up 3.5% of all transactions, meaning the data is heavily imbalanced and needs special handling when building a model, 
# not just accuracy as a measure of success.

# Transaction amount is extremely skewed. Most transactions are small, but a few very large ones stretch the range all the way up to nearly 32,000. 
# This means transaction amount alone cannot be used as a simple cutoff to detect fraud, 
# and it should be log transformed before being used in a model.

# Fraud rate is not the same across product categories. Category C shows the highest fraud rate at 11 percent, 
# while category W shows the lowest at just over 2 percent. This difference was confirmed to be real and not due to chance using a statistical test.

# Fraud rate also differs by device type. Mobile transactions show a fraud rate of over 10 percent, more than double the desktop rate, 
# while transactions with no device data show the lowest fraud rate.

# Card1 does not always represent one individual card. Most cards behave normally, but a small group, about 1.5 percent of all cards, 
# show extremely high transaction counts that are not realistic for a single person, suggesting they are shared or default codes rather 
# than real individual cards.

# Fraud rate spikes sharply during early morning hours, roughly between 5 AM and 10 AM, reaching over 10 percent at the peak, 
# more than three times higher than the normal rate.

# Direct correlation between fraud and simple numeric columns like amount, time, and card number is very weak. 
# This does not mean these factors do not matter, it means fraud is driven by specific patterns and categories 
# rather than a simple straight line relationship, which is why grouping the data and testing it properly revealed much 
# stronger patterns than correlation alone could show.

# Across many columns, missing values were found to carry real meaning rather than being random gaps. In several cases, 
# whether a value was present or missing showed a clear and measurable difference in fraud rate, 
# which is why those columns were not filled in with guessed values during cleaning.
```
