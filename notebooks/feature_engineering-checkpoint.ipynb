```python
import pandas as pd

#import database
df = pd.read_parquet('sentinel_features_eda.parquet')

# ============================================================
# Presence Flags - Identity Column Groups
# These flags do NOT represent fraud or genuine directly.
# They capture whether specific groups of identity columns 
# have real data (present) or are missing, for each transaction.
# The link to fraud is a SEPARATE finding, checked afterward.
# ============================================================

# Group 1: id_07, id_08, id_21to id_27
# Some columns in this group are text-based (filled with 'unknown' 
# during cleaning), so presence check excludes 'unknown' as well as NaN.
# Flag = 1 if AT LEAST ONE column in this group has real data (.any)
id_group1_cols = ['id_07', 'id_08', 'id_21', 'id_22', 'id_23', 'id_24', 'id_25', 'id_26', 'id_27']
presence_check = df[id_group1_cols].apply(lambda col: col.notna() & (col != 'unknown'))
df['id_group1_present'] = presence_check.any(axis=1).astype(int)

# Verify
print(df.groupby('id_group1_present')['isFraud'].mean() * 100)
print(df['id_group1_present'].value_counts())

#Group 1 : present rate - 8.24% vs missing rate - 3.46%
```

```python
# Group 2 id columns: broad identity cluster (mostly text-based columns)
# Flag = 1 if AT LEAST ONE column in this group has real data (.any)
id_group2_cols = ['id_01', 'id_02', 'id_05', 'id_06', 'id_11', 'id_12', 'id_15', 'id_17', 'id_19', 'id_20', 
                  'id_28', 'id_29', 'id_31', 'id_35', 'id_36', 'id_37', 'id_38']
presence_check = df[id_group2_cols].apply(lambda col: col.notna() & (col != 'unknown'))
df['id_group2_present'] = presence_check.any(axis=1).astype(int)

# Verify
print(df.groupby('id_group2_present')['isFraud'].mean() * 100)
print(df['id_group2_present'].value_counts())

#Group 2 : present rate - 7.85% vs missing rate - 2.09%
```

```python
# Group 3: id_03, id_04, id_09, id_10
# Flag = 1 if AT LEAST ONE column in this group has real data (.any)
id_group3_cols = ['id_03', 'id_04', 'id_09', 'id_10']
presence_check = df[id_group3_cols].apply(lambda col: col.notna() & (col != 'unknown'))
df['id_group3_present'] = presence_check.any(axis=1).astype(int)

# Verify
print(df.groupby('id_group3_present')['isFraud'].mean() * 100)
print(df['id_group3_present'].value_counts())

#Group 3 : present rate - 10.44% vs missing rate - 2.49%
```

```python
# Group 4: id_14, id_30, id_32, id_33, id_34
# Flag = 1 if AT LEAST ONE column in this group has real data (.any)
id_group4_cols = ['id_14', 'id_30', 'id_32', 'id_33', 'id_34']
presence_check = df[id_group4_cols].apply(lambda col: col.notna() & (col != 'unknown'))
df['id_group4_present'] = presence_check.any(axis=1).astype(int)

# Verify
print(df.groupby('id_group4_present')['isFraud'].mean() * 100)
print(df['id_group4_present'].value_counts())

#Group 4 : present rate - 4.50% vs missing rate - 3.34%
```

```python
# Insight — Identity Column Presence Flags

# Built four separate flags to capture whether specific groups of identity columns are present or missing for each transaction, 
# instead of treating all 38 identity columns as individual signals. This was based on earlier testing that showed columns 
# clustering into four distinct groups, each with different fraud rate patterns.

# Group 1, covering id_07, id_08, and several related columns, shows a fraud rate of 8.24 percent when present versus 3.46 percent 
# when missing. 
# Group 2, the broadest group covering most general identity columns, shows 7.85 percent(present) versus 2.09 percent(missing). 
# Group 3, covering id_03, id_04, id_09, and id_10, shows the strongest signal at 10.44 percent(present) versus 2.49 percent(missing). 
# Group 4 shows the weakest signal, 4.50(present) percent versus 3.34 percent(missing).

# These four flags now give the model a clear, verified signal about whether specific types of identity checks were triggered for each transaction, 
# without losing the meaningful differences between the groups.
```

```python
# ============================================================
# Presence Flags - D-Column Groups
# D-columns split into two groups showing OPPOSITE fraud-rate 
# directions when tested earlier. Flags capture presence pattern 
# only - the fraud direction is a separate, already-confirmed finding.
# ============================================================

# Group A: D6, D7, D8, D9, D12, D13, D14
# Earlier testing showed fraud rate is HIGHER when these are present.
# Flag = 1 if AT LEAST ONE column in this group has real data (.any)
D_groupA_cols = ['D6', 'D7', 'D8', 'D9', 'D12', 'D13', 'D14']
presence_check = df[D_groupA_cols].apply(lambda col: col.notna() & (col != 'unknown'))
df['D_groupA_present'] = presence_check.any(axis=1).astype(int)

# Verify
print(df.groupby('D_groupA_present')['isFraud'].mean() * 100)
print(df['D_groupA_present'].value_counts())

```

```python
# Group B: D1, D2, D3, D4, D5, D10, D11, D15
# Earlier testing showed fraud rate is HIGHER when these are missing.
# D1 is present in almost every row, so .any() would make this flag 
# meaningless (nearly always 1). Using .all() instead:
# Flag = 1 only if EVERY column in this group has real data
D_groupB_cols = ['D1', 'D2', 'D3', 'D4', 'D5', 'D10', 'D11', 'D15']
presence_check = df[D_groupB_cols].apply(lambda col: col.notna() & (col != 'unknown'))
df['D_groupB_present'] = presence_check.all(axis=1).astype(int)

# Verify
print(df.groupby('D_groupB_present')['isFraud'].mean() * 100)
print(df['D_groupB_present'].value_counts())
```

```python
# Insight — D-Column Group Flags

# Built two separate presence flags for the D-columns, based on earlier testing that showed they split into two groups 
# with opposite fraud patterns rather than behaving the same way.

# Group A, covering D6, D7, D8, D9, D12, D13, and D14, shows a much higher fraud rate when the data is present, 9.15 percent, 
# compared to only 2.08 percent when missing. This flag was built using an any-present check, since one column 
# being available is enough to represent that this pattern was triggered.

# Group B, covering D1, D2, D3, D4, D5, D10, D11, and D15, shows the opposite pattern, a higher fraud rate when data is missing, 3.94 percent, 
# compared to 2.11 percent when everything is present. This flag needed an all-present check instead, since one of these columns, D1, 
# is filled in for almost every transaction. Using an any-present check here would have made the flag show as present for nearly every row, 
# hiding the real pattern.

# Keeping these as two separate flags, instead of combining all D-columns into one, preserves both signals correctly instead of letting them 
# cancel each other out.
```

```python
# ============================================================
# Presence Flags - M-Column Groups
# Confirmed earlier(data cleaning file): M1/M2/M3 move together perfectly, M7/M8/M9 
# are near-perfectly correlated (13 row exception). M4, M5, M6 
# behave individually - M6 strongest signal, M4 opposite direction.
# ============================================================

# M1-M3: perfectly correlated, one flag represents all three
M_group1_cols = ['M1', 'M2', 'M3']
presence_check = df[M_group1_cols].apply(lambda col: col.notna() & (col != 'unknown'))
df['M_group1_present'] = presence_check.any(axis=1).astype(int)

#verify
print(df.groupby('M_group1_present')['isFraud'].mean() * 100)
print(df['M_group1_present'].value_counts())
```

```python
# M7-M9: near-perfectly correlated (13 exceptions out of 590,540), 
# treated as one group using the same any-present logic as M1-M3
M_group2_cols = ['M7', 'M8', 'M9']
presence_check = df[M_group2_cols].apply(lambda col: col.notna() & (col != 'unknown'))
df['M_group2_present'] = presence_check.any(axis=1).astype(int)
print(df.groupby('M_group2_present')['isFraud'].mean() * 100)
print(df['M_group2_present'].value_counts())
```

```python
# M6: standalone flag, strongest individual M-column signal found earlier
df['M6_present'] = (df['M6'].notna() & (df['M6'] != 'unknown')).astype(int)

print(df.groupby('M6_present')['isFraud'].mean() * 100)
# (fraud rate: 2.06% present vs 7.07% missing - higher when MISSING)
```

```python
# M4: standalone flag.
df['M4_present'] = (df['M4'].notna() & (df['M4'] != 'unknown')).astype(int)

print(df.groupby('M4_present')['isFraud'].mean() * 100)
# (fraud rate: 4.99% present vs 1.86% missing - higher when PRESENT)
```

```python
# M5: standalone flag.
df['M5_present'] = (df['M5'].notna() & (df['M5'] != 'unknown')).astype(int)

print(df.groupby('M5_present')['isFraud'].mean() * 100)
# (fraud rate: 3.15% present vs 3.73% missing - higher when PRESENT)
```

```python
# Insight — M-Column Presence Flags

# Built flags for the M-columns based on earlier testing that showed some move together and others behave completely on their own.

# M1, M2, and M3 were confirmed to always appear or go missing together, so they were combined into one single flag. M7, M8, and M9 
# behave almost identically, with only a tiny number of exceptions, so they were also combined into one flag.

# M4, M5, and M6 were kept as separate individual flags, since they showed real differences in direction from each other. 
# M4 shows a higher fraud rate when present, 4.99 percent versus 1.86 percent when missing. M6 shows the opposite pattern, 
# a much stronger signal, 2.06 percent when present versus 7.07 percent when missing, meaning M6 being missing is actually the bigger red flag. 
# M5 showed the weakest, most balanced pattern of the three, so it was kept separate rather than forced into either direction.

# This approach avoids blending columns that behave differently into one confusing signal, keeping each real pattern clear and usable on its own.
```

```python
# addr1/addr2: strongest non-identity signal found earlier
# (fraud rate: 11.18% missing vs 2.51% present - higher when MISSING)
addr_cols = ['addr1', 'addr2']
df['addr_present'] = (df[addr_cols].notna() & (df[addr_cols] != 'unknown')).all(axis=1).astype(int)

print(df.groupby('addr_present')['isFraud'].mean() * 100)

# (fraud rate: 2.45% present vs 11.78% missing - higher when MISSING)
```

```python
# email_mismatch: 1 if both P_emaildomain and R_emaildomain are real values
# AND they differ from each other. 0 if either is missing/unknown, or if
# they match (same domain).

import numpy as np

p_known = df['P_emaildomain'].notna() & (df['P_emaildomain'] != 'unknown')
r_known = df['R_emaildomain'].notna() & (df['R_emaildomain'] != 'unknown')
both_known = p_known & r_known
is_match = both_known & (df['P_emaildomain'] == df['R_emaildomain'])
is_mismatch = both_known & (df['P_emaildomain'] != df['R_emaildomain'])

df['email_status'] = np.select(
    [is_match, is_mismatch],
    ['Match', 'Mismatch'],
    default='Cannot_Compare'
)

print(df.groupby('email_status')['isFraud'].mean() * 100)
print(df['email_status'].value_counts())

#INSIGHTS

# Built a three-category feature comparing purchaser and recipient email domains, since the earlier two-category version 
# blended genuinely different situations together and produced a confusing, misleading result.

# The three categories show clearly different fraud rates. When both emails are known and match, fraud rate is 9.65 percent, 
# the highest of the three. When both are known but different, fraud rate drops to 2.87 percent. When one or both emails are missing, 
# fraud rate is 2.17 percent, the lowest.

# This is the opposite of the original assumption. A mismatch was expected to be the stronger red flag, but the data shows the reverse. 
# A confirmed match may actually reflect situations where the real account holder is knowingly involved in the transaction, 
# such as first-party fraud or an authorized scam, where their own details stay fully consistent since no account takeover occurred. 
# This is different from account takeover, which usually introduces mismatches or missing data because a fraudster is acting on 
# someone else's account.
```

```python
#Performinh on hot emmcoding on email_status
email_dummies = pd.get_dummies(df['email_status'],prefix='email')
df = pd.concat([df,email_dummies],axis=1)
print(df[['email_status', 'email_Match', 'email_Mismatch', 'email_Cannot_Compare']].head(10))
```

```python
# Cheking hour fraud rate
print("Frad rate by hour:")
print(df.groupby('hour')['isFraud'].mean()*100)
```

```python
# Flagging 4am to 10am fraud rate.
df['is_high_risk_hour'] = ((df['hour'] >= 4) & (df['hour'] <= 10)).astype(int)

print(df['is_high_risk_hour'].value_counts())
print(df.groupby('is_high_risk_hour')['isFraud'].mean() * 100)
```

```python
# daily_velocity: how many transactions did this SAME card make on this 
# SAME day. Captures rapid repeated activity, similar to card-testing 
# transform() attaches the count back to every original row, unlike a 
# plain groupby which would collapse rows into a summary table.
df['daily_velocity'] = df.groupby(['card1', 'day'])['TransactionID'].transform('count')

print(df[['card1', 'day', 'daily_velocity']].head(20))

# Verify - check fraud rate for different velocity levels
print(df.groupby(pd.cut(df['daily_velocity'], bins=[0, 1, 5, 20, 100, 20000]))['isFraud'].mean() * 100)
```

```python
# card_avg_amount: this card's average transaction amount, attached to 
# every row for that card. Uses transform() so every transaction knows 
# its own card's typical spending level, not just a summary table.
df['card_avg_amount'] = df.groupby('card1')['TransactionAmt'].transform('mean')

# Checking difference between TransactionAmt and card_avg_amount
df['amt_difference'] = df['TransactionAmt'] - df['card_avg_amount']
print(df[['card1', 'TransactionAmt', 'card_avg_amount','amt_difference']].head(10))

# Diving fraud rate by amount difference bins.
print(df.groupby(pd.cut(df['amt_difference'], bins=[-5000,-1000,-500,-100,0, 50, 100, 500, 1000, 5000,20000]))['isFraud'].mean() * 100)
# Cheking Txn count between 5000 and 20000
print(df.groupby(pd.cut(df['amt_difference'], bins=[5000,20000]))['isFraud'].count())

# Insight — Daily Velocity

# Fraud rate is not simply higher with more transactions per card per day. It rises from 2.14 percent at a 
# single transaction to a peak of 4.13 percent at 6 to 20 transactions, then drops back down, falling to just 1.62 
# percent for cards with over 100 transactions in a day. This matches an earlier finding that very high-volume cards 
# are likely shared or default codes rather than real individual fraud activity, not a stronger fraud signal on their own.

# Insight — Amount vs Card Average

# Compared each transaction's amount to that same card's typical average spending. Fraud rate is lowest when a transaction is close to or 
# below the card's usual average, around 2.5 to 3 percent. It rises for moderate overspending, peaking around 6.3 percent 
# when a transaction is 500 to 1000 above the card's typical amount. One extreme bucket(5000, 20000) showed a much higher rate, but it 
# only had 8 transactions behind it, too small to trust, so it was not treated as a real finding.
```

```python
# Changing TranscationAmt into log.
df['log_amount'] = np.log1p(df['TransactionAmt'])

print(f"Original skew: {df['TransactionAmt'].skew()}")
print(f"Log-transformed skew: {df['log_amount'].skew()}")

# Insight — Log Transform of Transaction Amount

# TransactionAmt showed a skewness of 14.37, meaning the data is heavily lopsided, with most transactions small and a few extreme, 
# very large ones stretching the distribution far to one side. This kind of imbalance can make it harder for a model to learn patterns properly, 
# since the extreme values can dominate and distort what the model treats as "normal."

# Applying a log transform compresses large values much more than small ones, pulling the distribution closer to a normal, 
# symmetric shape. After the transform, skewness dropped from 14.37 to 0.49, a dramatic improvement, 
# confirming the data is now much better suited for modeling.

# log1p was used instead of a plain log, since it calculates log of the value plus one. 
# This avoids any issue with taking the log of zero, which is mathematically undefined, 
# even though this specific column's minimum value was safely above zero.
```
