```python
import pandas as pd

#Importing IEEE CIS dataset.
df = pd.read_parquet('ieee_merged_raw.parquet')  ##Merged train_transaction and train_identity csv and convereted into parquet data, since
                                                #the csv dataset is too large for system to run. (converstion steps in - csv_to_parquet_converstion file.)

#Cheking Shape of Dataset.
print("Rows and Columns of dataset:")
print(f"{df.shape}\n")                         ## There are rows - 590540 and columns - 434)
print('*' * 100)

#Cheking the Datatypes.
print("\nThe datatypes are:")
print(f"{df.dtypes}\n")                         ##It shows datatype of each column.
print('*' * 100)

#Cheking count of columns for each data type.
print(f"\n{df.info()}\n")                       ###dtypes: float16(354), float32(45), int16(1), int32(2), int8(1), object(31).
print('*' * 100)

#Cheking Duplicate Rows.
print("\nDuplicate Rows:")
print(f"{df.duplicated().sum()}\n")              ##No duplicate rows present.
print('*' * 100)
```

```python
#Cheking Null Count and Percentage.
print("\nNull Values Count:")
null_count = df.isnull().sum()                     ##Overall 414 columns out of 434 have nulls.
null_percent = null_count/len(df)*100
# Filter to only columns WITH nulls
null_count_filtered = null_count[null_count > 0].sort_values(ascending=False)
null_percent_filtered = null_percent[null_percent > 0].sort_values(ascending=False)
print(f"{null_count_filtered}\n")
print('*' * 100)
print("Null Values Percentage\n")
print(f"{null_percent_filtered}\n")
```

```python
#Cheking columns with zero nulls.
non_null_count = df.notna().sum()
non_null_percent = non_null_count/len(df) * 100
non_null_count_filtered = non_null_count[non_null_count > 0]
non_null_percent_filtered = non_null_percent[non_null_percent == 100]
print(f"Non Null Percent: {non_null_percent_filtered}")

# Select only columns that have missing values (less than 100% complete)
cols_with_nulls = non_null_percent[non_null_percent < 100].index.tolist()

print(f"Skipping 20 fully populated columns. Running missingness check on remaining {len(cols_with_nulls)} columns.")
```

```python
#Cheking Fraud rate for id_01 to id_38 columns.
print("Fraud Rate of id columns\n")
id_columns = ['id_01','id_02','id_03','id_04','id_05','id_06','id_07','id_08','id_09','id_10','id_11','id_12','id_13','id_14','id_15','id_16',
              'id_17','id_18','id_19','id_20','id_21','id_22','id_23','id_24','id_25','id_26','id_27','id_28','id_29','id_30','id_31','id_32',
              'id_33','id_34','id_35','id_36','id_37','id_38']

for col in id_columns:
    fraud_present = df[df[col].notna()]['isFraud'].mean() * 100
    fraud_missing = df[df[col].isna()]['isFraud'].mean() * 100
    count_present = df[col].notna().sum()

    print(f"{col}: fraud when present = {fraud_present:.2f}%, fraud when missing = {fraud_missing:.2f}%, count present = {count_present}")
```

```python
# Correlation between id columns
id_columns = ['id_01','id_02','id_03','id_04','id_05','id_06','id_07','id_08','id_09','id_10','id_11','id_12','id_13','id_14','id_15','id_16',
              'id_17','id_18','id_19','id_20','id_21','id_22','id_23','id_24','id_25','id_26','id_27','id_28','id_29','id_30','id_31','id_32',
              'id_33','id_34','id_35','id_36','id_37','id_38']

## Create a dataframe of True/False (as 0/1) showing whether each id_ column is present, per row
presence_df = df[id_columns].notna().astype(int)

# Correlation matrix - shows how similarly each pair of columns' missingness moves together
presence_corr = presence_df.corr()

print(presence_corr)

# Insight — id_ Column Correlation Analysis

# Checked whether the missing/present pattern across all 38 identity columns (id_01 to id_38) actually represents independent signals, 
# or if many of them are just duplicating the same underlying pattern. Built a correlation matrix on presence (1) vs missing (0) for every column, 
# rather than assuming similarity from matching counts alone.

# Finding:the columns aren't 38 independent signals, they cluster into roughly 4 large groups where columns move together almost perfectly 
# (correlation 0.95 to 1.0):

# 1. A large group (id_01, id_02, id_05, id_06, id_11, id_12, id_15, id_17, id_19, id_20, id_28, id_29, id_31, id_35-38) — 
#    likely tied to general device/browser fingerprinting, triggered together on most transactions with identity data
# 2. A smaller, high-fraud-rate group (id_07, id_08, id_21-27) — the one first discovered with an 8%+ fraud rate when present, versus 3.46% when missing
# 3. A mid-size group (id_03, id_04, id_09, id_10) — strongly related to each other (correlation 0.93) but not perfectly identical
# 4. Another group (id_14, id_30, id_32, id_33, id_34) — moderate present-count, lower fraud-rate lift than the id_07/08 group
```

```python
#Cheking Fraud rate for card columns.
print("Fraud Rate Analysis for Card Columns\n")
card_columns = ['card1', 'card2', 'card3', 'card4', 'card5', 'card6']

for col in card_columns:
    # Fraud rate when column is present
    fraud_present = df[df[col].notna()]['isFraud'].mean() * 100
    count_present = df[col].notna().sum()
    
    # Check missing count to avoid dividing by zero
    count_missing = df[col].isna().sum()
    if count_missing > 0:
        fraud_missing = df[df[col].isna()]['isFraud'].mean() * 100
        missing_str = f"{fraud_missing:.2f}%"
    else:
        missing_str = "N/A (No missing values)"

    print(f"{col} -> fraud when presen: {fraud_present:.2f}% (count present: {count_present}) | fraud when missing: {missing_str}")


# 1. Card1 has no null values.
# 2. There is a fraud rate change in card2 to card6 when null values are missing. Only in card2 and card5 the fraud rate increased 
# when nulls were missing not too high but there is a small increase.
```

```python
#Cheking Fraud rate for addr and dist columns.
print("Fraud Rate Analysis for addr and dist Columns\n")
columns = ['addr1','addr2','dist1','dist2']

for col in columns:
    fraud_present = df[df[col].notna()]['isFraud'].mean() * 100
    fraud_missing = df[df[col].isna()]['isFraud'].mean() * 100
    count_present = df[col].notna().sum()

    print(f"{col}: fraud when present = {fraud_present:.2f}%, fraud when missing = {fraud_missing:.2f}%, count present = {count_present}")


# 1. There is high jump in fraud rate in addr1 and addr2 when nulls are missing. The fraud rate went form 2.46% to 11.78% which is approx 5x increase.
# 2. In dist1 the fraud rate went up a bit when missing and in dist2 the fraud rate decreased a lot from 9.92% to 3.06% when missing.
```

```python
#Cheking Fraud rate for D columns.
print("Fraud Rate Analysis for D Columns\n")
columns = ['D1','D2','D3','D4','D5','D6','D7','D8','D9','D10','D11','D12','D13','D14','D15']

for col in columns:
    fraud_present = df[df[col].notna()]['isFraud'].mean() * 100
    fraud_missing = df[df[col].isna()]['isFraud'].mean() * 100
    count_present = df[col].notna().sum()

    print(f"{col}: fraud when present = {fraud_present:.2f}%, fraud when missing = {fraud_missing:.2f}%, count present = {count_present}")


# Checked all 15 D-columns for whether being present or missing relates to fraud rate, instead of assuming they all behave the same way.

# (D6, D7, D8, D9, D12, D13, D14):
# Fraud rate jumps to 10.45% – 14.88% when present (vs. 2.5% – 2.7% when missing).
# D7 shows the highest risk overall at 14.88% when present.

# (D2, D3, D10, D11, D15):
# Fraud rate increases to 4.20% – 5.21% when missing (vs. 1.96% – 3.27% when present).
# D11 shows the biggest increase, jumping from 1.96% to 5.21% when missing.

# Identical Pattern (D8 & D9):
# Exact same present count (74,926) and identical fraud rates (10.45% present / 2.49% missing), indicating they move together.
# No Significant Impact (D1, D4, D5):
# Fraud rate stays almost flat (3.16% – 3.88%) whether present or missing.
```

```python
# Check missingness fraud rate across all M columns
m_columns = [f'M{i}' for i in range(1, 10)]

for col in m_columns:
    fraud_present = df[df[col].notna()]['isFraud'].mean() * 100
    fraud_missing = df[df[col].isna()]['isFraud'].mean() * 100
    count_present = df[col].notna().sum()

    print(f"{col}: fraud when present = {fraud_present:.2f}%, fraud when missing = {fraud_missing:.2f}%, count present = {count_present}")


# M1,M2,M3 have same fraud rate of 1.99% when present and 5.28% when missing.
# M&,M8,M9 have same fraud rate of 1.97% when present and 4.58% when missing.
# M6 has the highest increase in fraud rate when missing it went from 2.06% to 7.07%.
# M4 the fraud rate decreased when missing is present went from 4.99% to 1.86%.
# M5 the fraud rate increased slightly.
```

```python
# Final Insight — Missingness Across the Entire Dataset

# 1. Tested whether missing values carry meaningful fraud signal across every major column group in the dataset — card details, address, distance, 
#    identity columns, time-delta columns, match columns, and Vesta engineered features.
# 2. Found the same core pattern repeating throughout: missingness is rarely random. In most column groups, whether a value is present or missing 
#    correlates with a real, measurable difference in fraud rate — sometimes higher when present, sometimes higher when missing, depending on the column.
# 3. The strength of this signal varies a lot — some columns show dramatic, reliable differences (like addr1/addr2, the id_07 cluster, and V170), 
#    while others show small or borderline effects, and a few dramatic-looking numbers turned out to come from very small sample sizes 
#    and shouldn't be trusted at face value.
# 4. Because of this, no column with real missingness was filled with an average, most common value, or guessed category. 
#    Filling any of them would have erased a pattern that may genuinely help detect fraud.
# 5. This dataset-wide pattern is the main reason the cleaning approach for this project leaves missing values as true missing data 
#    wherever they show a meaningful signal, rather than applying one blanket imputation rule to everything.
```

## Dataset Cleaning Summary

### Structure
1. After merging train_transaction and train_identity on TransactionID (left join): 
   590,540 rows, 434 columns.
2. TransactionID confirmed as primary key -- no duplicates, no nulls.
3. No duplicate rows found anywhere in the merged dataset.
4. isFraud confirmed as binary target --  only 0 and 1 values present.

### Business Rule Validation
5. TransactionAmt minimum value is 0.25 --  confirms no zero or negative 
   transaction amounts exist.

### Missing Data Investigation
6. Null percentages range widely across columns , from as low as 0.002% 
   to as high as 99% missing.
7. Tested whether missing values carry meaningful fraud signal across every 
   major column group — card details, address, distance, identity columns, 
   time-delta columns, match columns, and Vesta engineered features.
8. Found the same pattern repeating throughout: missingness is rarely random. 
   In most column groups, whether a value is present or missing correlates 
   with a real, measurable difference in fraud rate , sometimes higher when 
   present, sometimes higher when missing, depending on the column.
9. Some column groups were found to be fully or near-fully redundant with 
   each other (e.g. M1/M2/M3, id_35-38) — confirmed using presence 
   correlation and crosstabs before deciding on flag strategy.

### Cleaning Decisions
10. All genuinely categorical columns (confirmed by checking actual dtype, 
    not assumed by column-name range) filled with 'Unknown' , includes 
    card4, card6, DeviceType, P_emaildomain, R_emaildomain, M1-M9, and the 
    id_ columns confirmed as object dtype (id_12, id_15, id_16, id_23, 
    id_27-31, id_33-38).
11. All numeric columns with real missingness left as true NaN ,not filled 
    with mean, median, or mode. Filling them would erase the fraud signal 
    found in point 8, and would distort statistics and model training later.
12. Corrected an initial assumption that all id_12-38 columns were 
    categorical — dtype check revealed 12 of them (id_13, id_14, id_17-22, 
    id_24-26, id_32) are actually numeric and were left as NaN instead.

### Technical Notes
13. Numeric columns were downcast to smaller memory-efficient types 
    (int8/16/32, float16/32) using a custom reduce_mem_usage function 
    before merging, since the full dataset would not fit in memory at 
    default pandas dtypes on this machine.
14. Final cleaned dataset saved as: sentinel_features_cleaned.parquet
